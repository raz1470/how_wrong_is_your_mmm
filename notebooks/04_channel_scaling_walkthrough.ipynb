# Research study: does the collinearity story hold at realistic channel counts?

**Question:** Notebooks 01–03 built the collinearity diagnostic and budget-phasing fix using a 3-channel TV/Meta/Search toy example. Real MMMs often run 10–20+ channels (TV, paid social, paid search, display, OOH, radio, affiliates, and so on). Does the story hold up at realistic scale?

- **Section 1** — baseline unreliability vs channel count: holding correlation fixed, how does marginal return band width change as the number of channels grows from 5 to 20?
- **Section 2** — phasing benefit vs channel count: does budget phasing still meaningfully reduce that unreliability at higher channel counts, or does the same phasing effort dilute once it's spread across more channels?

All channels share one latent demand signal (the demand-driven DGP from notebook 01), so more channels competing for the same signal is expected to make individual channel effects harder to identify — this notebook quantifies how much harder, and whether phasing keeps up.

`BudgetPhaser` and `CollinearityDiagnostic` are already N-channel generalised (session 4); nothing in the package needed to change for this notebook.


```python
FAST_MODE = False  # set False for publication-quality runs (slower)

import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from how_wrong_is_your_mmm import Blackout, BudgetPhaser, CollinearityDiagnostic
from how_wrong_is_your_mmm._dgp import simulate_spend

N_SIMS = 10 if FAST_MODE else 50
N_PHASING_SEEDS = 1 if FAST_MODE else 10

print(
    f"FAST_MODE={FAST_MODE}  |  N_SIMS={N_SIMS}  |  N_PHASING_SEEDS={N_PHASING_SEEDS}"
)
```

```python
# ── Fixed params ───────────────────────────────────────────────────────────
CORRELATION = 0.7  # realistic inter-channel correlation, same as notebooks 02/03
N_HISTORY = 208  # 4 years of weekly data
PLAN_WEEKS = 52  # 1 year plan
MAX_WEEKLY_DEV_PCT = 40.0
CHANNEL_COUNTS = [5, 10, 15, 20]
BASELINE_DRAW_SEEDS = (
    5  # independent (marginal return, spend) draws averaged per N, Section 1
)
PHASING_DRAW_SEEDS = 8  # independent draws averaged per N, Section 2 (bumped from 3 to resolve whether the N=15 vs N=20 dip is real or noise)


def make_channels_and_marginal_returns(n_channels, seed=42):
    """Generic channel names + reproducible marginal_returns.

    Marginal returns are drawn from Uniform(0.2, 0.6) for a given seed, so results
    are reproducible and comparable across channel counts and draws without
    hand-picking values (avoids marginal return choice as a confound).
    """
    channels = [f"ch{i:02d}" for i in range(1, n_channels + 1)]
    rng = np.random.default_rng(seed)
    marginal_returns = {ch: float(rng.uniform(0.2, 0.6)) for ch in channels}
    return channels, marginal_returns


# ── Band width helper (same definition as notebook 03) ───────────────────────
def band_width(diag: CollinearityDiagnostic) -> pd.Series:
    """Return 90th–10th pct band width per channel from a fitted diagnostic."""
    return diag.results_.groupby("channel")["estimated_marginal_return"].apply(
        lambda x: float(x.quantile(0.9) - x.quantile(0.1))
    )


for n in CHANNEL_COUNTS:
    channels, marginal_returns = make_channels_and_marginal_returns(n)
    print(
        f"N={n:>2}  |  marginal return range [{min(marginal_returns.values()):.2f}, {max(marginal_returns.values()):.2f}]"
    )
```

## Section 1 — Baseline unreliability vs channel count

Fix correlation at 0.7 (the realistic level used throughout). For each channel count, generate 208 weeks of history and fit the diagnostic with no phasing.

A single random draw of marginal_returns and spend history is noisy relative to the size of effect we're looking for, so each channel count is averaged over `BASELINE_DRAW_SEEDS` independent draws (different marginal_returns and spend realisations, same correlation). This separates the true channel-count effect from single-draw luck — the same principle behind averaging phasing seeds in notebook 03.


```python
baseline_records = []
baseline_bw_pooled = {}  # pooled per-channel band widths across draws, for the boxplot

for n in CHANNEL_COUNTS:
    draw_means, draw_medians, draw_maxes = [], [], []
    pooled = []
    for draw in range(BASELINE_DRAW_SEEDS):
        channels, marginal_returns = make_channels_and_marginal_returns(
            n, seed=42 + draw
        )
        history = simulate_spend(
            n_obs=N_HISTORY, correlation=CORRELATION, channels=channels, seed=draw
        )

        diag = CollinearityDiagnostic(
            spend_df=history, true_marginal_returns=marginal_returns
        )
        diag.fit(n_sims=N_SIMS)
        bw = band_width(diag)

        draw_means.append(bw.mean())
        draw_medians.append(bw.median())
        draw_maxes.append(bw.max())
        pooled.extend(bw.to_numpy())

    baseline_bw_pooled[n] = np.array(pooled)
    baseline_records.append(
        {
            "n_channels": n,
            "mean_bw": float(np.mean(draw_means)),
            "median_bw": float(np.mean(draw_medians)),
            "max_bw": float(np.mean(draw_maxes)),
        }
    )
    print(
        f"N={n:>2}  |  mean_bw={np.mean(draw_means):.3f}  median_bw={np.mean(draw_medians):.3f}  max_bw={np.mean(draw_maxes):.3f}  (avg over {BASELINE_DRAW_SEEDS} draws)"
    )

df_baseline = pd.DataFrame(baseline_records)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Left: mean/median/max band width vs channel count
ax = axes[0]
ax.plot(
    df_baseline["n_channels"],
    df_baseline["mean_bw"],
    color="steelblue",
    marker="o",
    linewidth=2,
    label="mean",
)
ax.plot(
    df_baseline["n_channels"],
    df_baseline["median_bw"],
    color="mediumseagreen",
    marker="s",
    linewidth=2,
    label="median",
)
ax.plot(
    df_baseline["n_channels"],
    df_baseline["max_bw"],
    color="coral",
    marker="^",
    linewidth=2,
    label="max",
)
ax.set_xlabel("Number of channels")
ax.set_ylabel("Band width (90th–10th pct)")
ax.set_title(
    f"Marginal return band width vs channel count\n(correlation=0.7, no phasing, avg over {BASELINE_DRAW_SEEDS} draws)"
)
ax.set_xticks(CHANNEL_COUNTS)
ax.legend(fontsize=9)

# Right: pooled distribution of per-channel band widths at each N, across all draws
ax2 = axes[1]
box_data = [baseline_bw_pooled[n] for n in CHANNEL_COUNTS]
ax2.boxplot(box_data, tick_labels=[str(n) for n in CHANNEL_COUNTS])
ax2.set_xlabel("Number of channels")
ax2.set_ylabel("Band width (90th–10th pct)")
ax2.set_title(
    f"Per-channel band width distribution by channel count\n(pooled across {BASELINE_DRAW_SEEDS} draws)"
)

plt.suptitle(
    "Section 1: baseline unreliability vs channel count",
    fontsize=12,
    fontweight="bold",
    y=1.02,
)
plt.tight_layout()
plt.show()

print("\n── Key findings ──────────────────────────────────────────────")
for _, row in df_baseline.iterrows():
    print(
        f"  N={int(row['n_channels']):>2}  |  mean_bw={row['mean_bw']:.3f}  max_bw={row['max_bw']:.3f}"
    )
pct_change = (
    (df_baseline["mean_bw"].iloc[-1] - df_baseline["mean_bw"].iloc[0])
    / df_baseline["mean_bw"].iloc[0]
    * 100
)
print(
    f"\n→ Mean band width changes by {pct_change:+.1f}% from N={CHANNEL_COUNTS[0]} to N={CHANNEL_COUNTS[-1]}, holding correlation fixed."
)
print(
    "  (Theory: for an equicorrelated design, VIF per channel converges to a finite ceiling"
)
print(
    "   1/(1-corr) as N grows, rather than increasing without bound — so a modest, decelerating"
)
print("   increase is the expected shape, not a runaway one.)")
```

## Section 2 — Does phasing still help at higher channel counts?

Fix the plan length at 52 weeks and deviation amplitude at the package default (40%). For each channel count, run `BudgetPhaser` to find the recommended phasing amplitude, then compare band width before (history + unphased plan) and after (history + phased plan at the recommended alpha) — the same before/after comparison as notebook 03, now repeated across channel counts.

Averaged over `PHASING_DRAW_SEEDS` independent draws per channel count (fewer than Section 1, since each draw here runs a full grid search).


```python
phasing_records = []

for n in CHANNEL_COUNTS:
    draw_alphas, draw_unphased, draw_phased, draw_reductions = [], [], [], []

    for draw in range(PHASING_DRAW_SEEDS):
        channels, marginal_returns = make_channels_and_marginal_returns(
            n, seed=42 + draw
        )
        history = simulate_spend(
            n_obs=N_HISTORY, correlation=CORRELATION, channels=channels, seed=draw
        )
        history.index = pd.date_range("2019-01-07", periods=N_HISTORY, freq="W-MON")
        plan = simulate_spend(
            n_obs=PLAN_WEEKS,
            correlation=CORRELATION,
            channels=channels,
            seed=100 + draw,
        )
        plan.index = pd.date_range(
            history.index[-1] + pd.Timedelta(weeks=1), periods=PLAN_WEEKS, freq="W-MON"
        )

        # Baseline: history + unphased plan
        combined_unphased = pd.concat([history, plan])
        diag_unphased = CollinearityDiagnostic(
            spend_df=combined_unphased, true_marginal_returns=marginal_returns
        )
        diag_unphased.fit(n_sims=N_SIMS)
        bw_unphased = band_width(diag_unphased).max()

        # Run BudgetPhaser to find the recommended schedule
        phaser = BudgetPhaser(
            history_df=history,
            plan_df=plan,
            true_marginal_returns=marginal_returns,
            max_weekly_deviation_pct=MAX_WEEKLY_DEV_PCT,
        )
        phaser.fit(n_sims=N_SIMS, grid_steps=15, n_phasing_seeds=N_PHASING_SEEDS)
        rec = phaser.recommend()

        # Band width at the recommended schedule (history + phased plan)
        combined_phased = pd.concat([history, phaser.recommended_schedule_])
        diag_phased = CollinearityDiagnostic(
            spend_df=combined_phased, true_marginal_returns=marginal_returns
        )
        diag_phased.fit(n_sims=N_SIMS)
        bw_phased = band_width(diag_phased).max()

        reduction = (bw_unphased - bw_phased) / bw_unphased * 100
        draw_alphas.append(rec["alpha"])
        draw_unphased.append(bw_unphased)
        draw_phased.append(bw_phased)
        draw_reductions.append(reduction)

    phasing_records.append(
        {
            "n_channels": n,
            "best_alpha": float(np.mean(draw_alphas)),
            "max_bw_unphased": float(np.mean(draw_unphased)),
            "max_bw_phased": float(np.mean(draw_phased)),
            "reduction_pct": float(np.mean(draw_reductions)),
            "reduction_pct_std": float(np.std(draw_reductions)),
        }
    )
    print(
        f"N={n:>2}  |  alpha={np.mean(draw_alphas):.2f}  |  max_bw unphased={np.mean(draw_unphased):.3f}  phased={np.mean(draw_phased):.3f}  reduction={np.mean(draw_reductions):+.1f}% (±{np.std(draw_reductions):.1f}, {PHASING_DRAW_SEEDS} draws)"
    )

df_phasing = pd.DataFrame(phasing_records)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Left: max band width, unphased vs phased, by channel count
ax = axes[0]
width = 0.35
x = np.arange(len(CHANNEL_COUNTS))
ax.bar(
    x - width / 2,
    df_phasing["max_bw_unphased"],
    width,
    color="slategray",
    alpha=0.6,
    label="Unphased",
)
ax.bar(
    x + width / 2,
    df_phasing["max_bw_phased"],
    width,
    color="steelblue",
    alpha=0.9,
    label="Phased (recommended)",
)
ax.set_xticks(x)
ax.set_xticklabels([str(n) for n in CHANNEL_COUNTS])
ax.set_xlabel("Number of channels")
ax.set_ylabel("Max band width across channels")
ax.set_title(
    f"Phasing reduces max band width at every channel count tested\n(avg over {PHASING_DRAW_SEEDS} draws)"
)
ax.legend(fontsize=9)

# Right: % reduction vs channel count, with draw-to-draw std as error bars
ax2 = axes[1]
ax2.bar(
    x,
    df_phasing["reduction_pct"],
    yerr=df_phasing["reduction_pct_std"],
    capsize=4,
    color="mediumseagreen",
    alpha=0.85,
)
ax2.axhline(0, color="black", linewidth=1)
ax2.set_xticks(x)
ax2.set_xticklabels([str(n) for n in CHANNEL_COUNTS])
ax2.set_xlabel("Number of channels")
ax2.set_ylabel("Reduction in max band width (%)")
ax2.set_title(
    f"Phasing benefit (%) by channel count\n(error bars: std across {PHASING_DRAW_SEEDS} draws)"
)
for i, v in enumerate(df_phasing["reduction_pct"]):
    ax2.text(
        i,
        v + df_phasing["reduction_pct_std"].iloc[i] + 1,
        f"{v:.1f}%",
        ha="center",
        fontsize=10,
    )

plt.suptitle(
    "Section 2: phasing benefit across channel counts",
    fontsize=12,
    fontweight="bold",
    y=1.02,
)
plt.tight_layout()
plt.show()

print("\n── Key findings ──────────────────────────────────────────────")
for _, row in df_phasing.iterrows():
    print(
        f"  N={int(row['n_channels']):>2}  |  best alpha={row['best_alpha']:.2f}  |  reduction={row['reduction_pct']:+.1f}% (±{row['reduction_pct_std']:.1f})"
    )
print(
    f"\n→ Phasing reduced max band width at every channel count tested (all draws averaged)."
)
```

## Section 2b — Does Blackout's extra leverage hold at higher channel counts?

Notebook 02's per-channel deviation constraints showed `Blackout(max_dark_weeks_per_month=1)` producing a much bigger CV reduction than the package's default continuous +/-40% range, in the 3-channel toy example. Section 2 above showed channel count barely matters for the *size* of the standard phasing benefit. This section checks whether Blackout's extra leverage over a continuous range survives at higher channel counts too, or whether it's a small-channel-count artefact.

Same channel counts, draws, and settings as Section 2 above (`CHANNEL_COUNTS`, `N_SIMS`, `PHASING_DRAW_SEEDS`) — only the deviation shape (`max_weekly_deviation_pct`) changes between the two series.

```python
blackout_shapes = {
    "dev=40 (Section 2)": MAX_WEEKLY_DEV_PCT,
    "Blackout(cap=1)": Blackout(max_dark_weeks_per_month=1),
}

blackout_phasing_records = []

for n in CHANNEL_COUNTS:
    for label, dev in blackout_shapes.items():
        draw_reductions = []
        for draw in range(PHASING_DRAW_SEEDS):
            channels, marginal_returns = make_channels_and_marginal_returns(
                n, seed=42 + draw
            )
            history = simulate_spend(
                n_obs=N_HISTORY, correlation=CORRELATION, channels=channels, seed=draw
            )
            history.index = pd.date_range("2019-01-07", periods=N_HISTORY, freq="W-MON")
            plan = simulate_spend(
                n_obs=PLAN_WEEKS,
                correlation=CORRELATION,
                channels=channels,
                seed=100 + draw,
            )
            plan.index = pd.date_range(
                history.index[-1] + pd.Timedelta(weeks=1),
                periods=PLAN_WEEKS,
                freq="W-MON",
            )

            combined_unphased = pd.concat([history, plan])
            diag_unphased = CollinearityDiagnostic(
                spend_df=combined_unphased, true_marginal_returns=marginal_returns
            )
            diag_unphased.fit(n_sims=N_SIMS)
            bw_unphased = band_width(diag_unphased).max()

            phaser = BudgetPhaser(
                history_df=history,
                plan_df=plan,
                true_marginal_returns=marginal_returns,
                max_weekly_deviation_pct=dev,
            )
            phaser.fit(n_sims=N_SIMS, grid_steps=15, n_phasing_seeds=N_PHASING_SEEDS)

            combined_phased = pd.concat([history, phaser.recommended_schedule_])
            diag_phased = CollinearityDiagnostic(
                spend_df=combined_phased, true_marginal_returns=marginal_returns
            )
            diag_phased.fit(n_sims=N_SIMS)
            bw_phased = band_width(diag_phased).max()

            draw_reductions.append((bw_unphased - bw_phased) / bw_unphased * 100)

        blackout_phasing_records.append(
            {
                "n_channels": n,
                "shape": label,
                "reduction_pct": float(np.mean(draw_reductions)),
                "reduction_pct_std": float(np.std(draw_reductions)),
            }
        )
        print(
            f"N={n:>2}  {label:20s}  reduction={np.mean(draw_reductions):+.1f}% "
            f"(±{np.std(draw_reductions):.1f}, {PHASING_DRAW_SEEDS} draws)"
        )

df_blackout_phasing = pd.DataFrame(blackout_phasing_records)
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
width = 0.35
x = np.arange(len(CHANNEL_COUNTS))

dev40_vals = df_blackout_phasing[df_blackout_phasing["shape"] == "dev=40 (Section 2)"][
    "reduction_pct"
].to_numpy()
dev40_std = df_blackout_phasing[df_blackout_phasing["shape"] == "dev=40 (Section 2)"][
    "reduction_pct_std"
].to_numpy()
blackout_vals = df_blackout_phasing[df_blackout_phasing["shape"] == "Blackout(cap=1)"][
    "reduction_pct"
].to_numpy()
blackout_std = df_blackout_phasing[df_blackout_phasing["shape"] == "Blackout(cap=1)"][
    "reduction_pct_std"
].to_numpy()

ax.bar(
    x - width / 2,
    dev40_vals,
    width,
    yerr=dev40_std,
    capsize=4,
    color="slategray",
    alpha=0.75,
    label="dev=40 (continuous range)",
)
ax.bar(
    x + width / 2,
    blackout_vals,
    width,
    yerr=blackout_std,
    capsize=4,
    color="mediumseagreen",
    alpha=0.9,
    label="Blackout(cap=1)",
)
ax.set_xticks(x)
ax.set_xticklabels([str(n) for n in CHANNEL_COUNTS])
ax.set_xlabel("Number of channels")
ax.set_ylabel("Reduction in max band width (%)")
ax.set_title(
    f"Blackout vs continuous range across channel counts\n"
    f"(error bars: std across {PHASING_DRAW_SEEDS} draws)"
)
ax.legend(fontsize=9)
for i in range(len(CHANNEL_COUNTS)):
    ax.text(
        i - width / 2,
        dev40_vals[i] + dev40_std[i] + 1,
        f"{dev40_vals[i]:.0f}%",
        ha="center",
        fontsize=9,
    )
    ax.text(
        i + width / 2,
        blackout_vals[i] + blackout_std[i] + 1,
        f"{blackout_vals[i]:.0f}%",
        ha="center",
        fontsize=9,
    )
plt.tight_layout()
plt.show()

print("\n── Key findings ──────────────────────────────────────────────")
for n in CHANNEL_COUNTS:
    row_dev = df_blackout_phasing[
        (df_blackout_phasing["n_channels"] == n)
        & (df_blackout_phasing["shape"] == "dev=40 (Section 2)")
    ].iloc[0]
    row_bo = df_blackout_phasing[
        (df_blackout_phasing["n_channels"] == n)
        & (df_blackout_phasing["shape"] == "Blackout(cap=1)")
    ].iloc[0]
    print(
        f"  N={n:>2}  |  dev=40: {row_dev['reduction_pct']:+.1f}%  |  "
        f"Blackout: {row_bo['reduction_pct']:+.1f}%"
    )
```

**Blackout's extra leverage holds at every channel count tested, and doesn't dilute with scale.** Using this section's own full settings (`n_sims=50`, `grid_steps=15`, 8 draws per channel count), Blackout consistently reduces max band width by roughly 2.5-2.8x more than the continuous +/-40% range at every N from 5 to 20: 52.9% (N=5), 56.0% (N=10), 50.8% (N=15), and 50.6% (N=20), against the continuous range's own 21.4%, 22.4%, 19.6%, and 18.2% at the same channel counts. Blackout's advantage doesn't narrow with scale here — if anything it holds roughly steady to slightly wider (about 2.5x at N=5 to about 2.8x at N=20) — showing no sign of collapsing. This is the same flat-across-channel-count pattern Section 1 and Section 2 already established for the baseline effect and the standard phasing benefit — channel count isn't the lever, correlation and deviation shape are.

## Section 3 — What the recommended schedule actually looks like (N=15 example)

Sections 1 and 2 reported summary statistics — band width, % reduction — across channel counts. This section makes it concrete for one representative case (N=15, roughly the middle of the realistic range most MMMs run) by showing the actual weekly spend schedule and the correlation structure before and after phasing.


```python
# ── Build one representative N=15 example (same draw=0 seeds as Section 2) ──
EXAMPLE_N = 15
example_channels, example_marginal_returns = make_channels_and_marginal_returns(
    EXAMPLE_N, seed=42
)
example_history = simulate_spend(
    n_obs=N_HISTORY, correlation=CORRELATION, channels=example_channels, seed=0
)
example_history.index = pd.date_range("2019-01-07", periods=N_HISTORY, freq="W-MON")
example_plan = simulate_spend(
    n_obs=PLAN_WEEKS, correlation=CORRELATION, channels=example_channels, seed=100
)
example_plan.index = pd.date_range(
    example_history.index[-1] + pd.Timedelta(weeks=1), periods=PLAN_WEEKS, freq="W-MON"
)

example_phaser = BudgetPhaser(
    history_df=example_history,
    plan_df=example_plan,
    true_marginal_returns=example_marginal_returns,
    max_weekly_deviation_pct=MAX_WEEKLY_DEV_PCT,
)
example_phaser.fit(n_sims=N_SIMS, grid_steps=15, n_phasing_seeds=N_PHASING_SEEDS)
example_rec = example_phaser.recommend()
example_phased = example_phaser.recommended_schedule_

print(
    f"N={EXAMPLE_N}  |  recommended alpha={example_rec['alpha']:.2f}  |  max_cv unphased vs phased: see Section 2 aggregate"
)
```

```python
# ── Weekly spend: original vs phased, 4 representative channels ─────────────
# Pick channels spread across the marginal return range (smallest, ~1/3, ~2/3, largest)
# rather than an arbitrary first-4 slice, so the example is representative.
sorted_channels = sorted(example_marginal_returns, key=example_marginal_returns.get)
n = len(sorted_channels)
pick_idx = [round(i * (n - 1) / 3) for i in range(4)]
example_picks = [sorted_channels[i] for i in pick_idx]

pick_colors = ["steelblue", "coral", "mediumseagreen", "goldenrod"]

fig, axes = plt.subplots(len(example_picks), 1, figsize=(13, 9), sharex=True)
for ax, ch, color in zip(axes, example_picks, pick_colors):
    ax.plot(
        example_plan.index,
        example_plan[ch] / 1_000,
        color="grey",
        alpha=0.6,
        linewidth=1.2,
        label="original plan",
    )
    ax.plot(
        example_phased.index,
        example_phased[ch] / 1_000,
        color=color,
        linewidth=1.5,
        label=f"phased (\u03b1={example_rec['alpha']:.2f})",
    )
    ax.set_ylabel("GBP k / week")
    ax.set_title(
        f"{ch} weekly spend (marginal return={example_marginal_returns[ch]:.2f})"
    )
    ax.legend(fontsize=8)

plt.suptitle(
    f"N={EXAMPLE_N} example: original plan vs phased prescription \u2014 monthly totals preserved\n"
    f"(4 representative channels, spread across the marginal return range)",
    fontsize=11,
    y=1.01,
)
plt.tight_layout()
plt.show()
```

```python
# ── Correlation matrix: before vs after phasing ──────────────────────────────
corr_before = example_plan.corr().to_numpy()
corr_after = example_phased.corr().to_numpy()

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
for ax, corr, title in zip(
    axes,
    [corr_before, corr_after],
    ["Before phasing (original plan)", "After phasing (recommended schedule)"],
):
    im = ax.imshow(corr, vmin=-1, vmax=1, cmap="RdBu_r")
    ax.set_title(title, fontsize=11)
    ax.set_xticks(range(EXAMPLE_N))
    ax.set_yticks(range(EXAMPLE_N))
    ax.set_xticklabels(example_channels, rotation=90, fontsize=6)
    ax.set_yticklabels(example_channels, fontsize=6)

fig.colorbar(im, ax=axes, shrink=0.8, label="Pearson correlation")
plt.suptitle(
    f"N={EXAMPLE_N}: inter-channel correlation before and after phasing",
    fontsize=12,
    fontweight="bold",
    y=1.02,
)
plt.show()

mean_corr_before = float(
    np.mean(
        [corr_before[i, j] for i in range(EXAMPLE_N) for j in range(i + 1, EXAMPLE_N)]
    )
)
mean_corr_after = float(
    np.mean(
        [corr_after[i, j] for i in range(EXAMPLE_N) for j in range(i + 1, EXAMPLE_N)]
    )
)
print(
    f"Mean pairwise correlation: {mean_corr_before:.3f} (before) \u2192 {mean_corr_after:.3f} (after)"
)
```

### CAC and ROI for the N=15 example

Section 2 reported the channel-count sweep in band-width terms. Here's the concrete N=15 example from Section 3 above, read off in CAC and ROI terms instead, for the channel with the widest unphased band — same illustrative `value_per_unit = £150` per acquisition as notebooks 01–03. This is one worked example, not a sweep; it's here to show the £ framing holds up at a realistic channel count, not just the 3-channel toy example.


```python
value_per_unit = 150.0

diag_unphased = CollinearityDiagnostic(
    spend_df=pd.concat([example_history, example_plan]),
    true_marginal_returns=example_marginal_returns,
)
diag_unphased.fit(n_sims=N_SIMS)

diag_phased = CollinearityDiagnostic(
    spend_df=pd.concat([example_history, example_phased]),
    true_marginal_returns=example_marginal_returns,
)
diag_phased.fit(n_sims=N_SIMS)

worst_channel = band_width(diag_unphased).idxmax()
unphased_summary = diag_unphased.summary(value_per_unit=value_per_unit)
phased_summary = diag_phased.summary(value_per_unit=value_per_unit)

print(f"Worst-case channel (widest unphased band, of {EXAMPLE_N}): {worst_channel}")
print()
for metric, label in [
    ("cac", "\u00a3 per acquisition"),
    ("roi", "\u00a3 back per \u00a3 spent"),
]:
    u_row = unphased_summary[unphased_summary["channel"] == worst_channel].iloc[0]
    p_row = phased_summary[phased_summary["channel"] == worst_channel].iloc[0]
    w_u = u_row[f"{metric}_p90"] - u_row[f"{metric}_p10"]
    w_p = p_row[f"{metric}_p90"] - p_row[f"{metric}_p10"]
    change_pct = 100 * (w_u - w_p) / w_u
    print(
        f"{metric.upper()} range width ({worst_channel}, {label}): "
        f"{w_u:.2f} -> {w_p:.2f}  ({change_pct:+.0f}%)"
    )
```

This repeats it with `Blackout(max_dark_weeks_per_month=1)` instead of the package's default continuous +/-40% range, reusing the same `example_history`/`example_plan`/`example_channels` — same two views as Section 3: the weekly spend prescription (now with dark weeks marked) and the before/after correlation matrix.

Same caveat as Section 2b: this is one worked example at one channel count, not a sweep, and switching every channel to `Blackout` is a bigger operational ask on the media-buying side than the continuous range Section 3 illustrates. But the pattern is now consistent across every context tested (3-channel toy example, time-to-benefit, and here at N=15): Blackout isn't a small-scale artefact, it keeps its edge as the story gets more realistic.

## What this means for practitioners

**Channel count barely matters for baseline unreliability; channel correlation is what drives it.** Averaged over 5 independent draws per channel count, mean band width was essentially flat from N=5 to N=20 (0.453 → 0.462, +1.9%), with a modest increase in the worst-case (max) band width across channels (0.537 → 0.574, +6.9%). This matches the theory: for channels sharing one latent demand signal at a fixed pairwise correlation, the per-channel variance inflation factor converges to a finite ceiling (1/(1-correlation)) as N grows, rather than increasing without bound. Practically: going from a 3-channel to a 20-channel MMM does not meaningfully worsen your model's reliability by itself. What matters, per notebook 03's Section 2, is how correlated your channels are with each other, not how many of them there are.

**Budget phasing keeps working at realistic channel counts, and the benefit doesn't decline with scale.** Averaged over 8 independent draws per channel count, phasing reduced max band width by +21.4% (N=5), +22.4% (N=10), +19.6% (N=15), and +18.2% (N=20) — a flat, consistent benefit across the whole range tested, well within each other's error bars. (An earlier pass with only 3 draws showed an apparent dip at N=20 relative to N=15; averaging over more draws resolved it as sampling noise rather than a real effect — a useful reminder that `max`-across-channels is a worst-case statistic that needs more averaging to read reliably as N grows.) The recommended phasing amplitude stayed at or near the maximum (alpha ≈ 0.96–1.00) at every channel count.

**Section 3 makes it concrete for a 15-channel example.** The recommended schedule uses maximum phasing (alpha=1.00) and cuts mean pairwise correlation among channels roughly by half, from 0.684 to 0.369. The weekly spend chart shows what that actually looks like for four representative channels spanning the marginal return range: the same media budget, redistributed week to week within each month, with monthly totals unchanged.

**Bottom line:** the diagnostic and the phasing fix both generalise cleanly beyond the 3-channel toy example used elsewhere in this project. No changes to `BudgetPhaser` or `CollinearityDiagnostic` were needed — they already handled N channels correctly (session 4). This notebook exists to prove that empirically, not to add new functionality.

## Section 4 — Is it the channel count or the correlation?

Sections 1-3 hold correlation fixed at 0.7 and vary the channel count. That
answers "does the story survive at 20 channels?" but not the question a
practitioner actually asks, which is *what should I worry about?*

This section sweeps both at once: channel count on one axis, inter-channel
correlation on the other, with nothing else moving. Marginal returns cycle
this project's own illustrative values (0.5 / 1.0 / 1.5), so at K=3 the setup
is literally the TV/Meta/Search scenario from notebook 02 and `overview.html`,
and no marginal-return draw varies across the grid. No phasing is applied
anywhere here: this measures the size of the problem, not the size of the fix.

The metric is the **coefficient of variation** of the recovered marginal
return, averaged across channels: how much a channel's estimate moves as a
percentage of the estimate itself. The mean is the right summary here rather
than the max, because the max over K channels is partly an extreme-value
artefact (more channels means more chances to draw an unlucky one), which
would show up as a channel-count effect that isn't really one. Both are
computed below so the difference is visible.


```python
# ── Section 4: correlation x channel count, no phasing ─────────────────────
PAGE_MARGINAL_RETURNS = [0.5, 1.0, 1.5]  # same illustrative values as notebook 02
SWEEP_CHANNEL_COUNTS = [3, 5, 10, 15, 20]
SWEEP_CORRELATIONS = [0.1, 0.3, 0.5, 0.7, 0.9]
SWEEP_SPEND_SEEDS = 10  # independent spend histories averaged per cell


def make_cycled_channels(n_channels):
    """Channel names + marginal returns cycling 0.5/1.0/1.5.

    Cycling rather than drawing at random removes the marginal-return draw as a
    confound: every cell in the grid sees the same set of channel economics, so
    any movement is attributable to correlation or channel count alone.
    """
    channels = [f"ch{i:02d}" for i in range(1, n_channels + 1)]
    marginal_returns = {
        ch: PAGE_MARGINAL_RETURNS[i % len(PAGE_MARGINAL_RETURNS)]
        for i, ch in enumerate(channels)
    }
    return channels, marginal_returns


sweep_records = []
for n in SWEEP_CHANNEL_COUNTS:
    channels, marginal_returns = make_cycled_channels(n)
    for corr in SWEEP_CORRELATIONS:
        means, maxes, actuals = [], [], []
        for spend_seed in range(SWEEP_SPEND_SEEDS):
            history = simulate_spend(
                n_obs=N_HISTORY, correlation=corr, channels=channels, seed=spend_seed
            )
            diag = CollinearityDiagnostic(
                spend_df=history, true_marginal_returns=marginal_returns
            ).fit(n_sims=N_SIMS)
            cv = diag.summary()["coef_of_variation"]
            means.append(cv.mean())
            maxes.append(cv.max())
            actuals.append(diag.actual_correlation)
        sweep_records.append(
            {
                "n_channels": n,
                "correlation": corr,
                "mean_cv_pct": 100 * float(np.mean(means)),
                "max_cv_pct": 100 * float(np.mean(maxes)),
                "actual_corr": float(np.mean(actuals)),
            }
        )
    print(f"K={n:>2} done")

df_sweep = pd.DataFrame(sweep_records)
grid_mean = df_sweep.pivot(index="n_channels", columns="correlation", values="mean_cv_pct")
grid_max = df_sweep.pivot(index="n_channels", columns="correlation", values="max_cv_pct")

print("\n── Mean CV %% (rows = channels, cols = correlation) ──")
print(grid_mean.round(1))
print("\n── Max CV %% (least reliable channel) ──")
print(grid_max.round(1))
```

```python
# ── What the grid says ─────────────────────────────────────────────────────
across = grid_mean.loc[3, 0.9] / grid_mean.loc[3, 0.1]
down = grid_mean.loc[20, 0.7] / grid_mean.loc[3, 0.7]

print("── Key findings ──────────────────────────────────────────────")
for n in SWEEP_CHANNEL_COUNTS:
    row = grid_mean.loc[n]
    print(f"  K={n:>2}  |  " + "  ".join(f"r={c}: {row[c]:4.1f}%" for c in SWEEP_CORRELATIONS))
print()
print(f"→ Holding channels at 3, going from correlation 0.1 to 0.9 multiplies")
print(f"  the average channel's CV by {across:.1f}x.")
print(f"→ Holding correlation at 0.7, going from 3 to 20 channels multiplies it")
print(f"  by {down:.2f}x.")
print()
print("  Correlation is the driver. Channel count is close to irrelevant for the")
print("  average channel, and the column-wise spread below shows how little it moves.")
print(f"  Per-column range across all channel counts:")
for c in SWEEP_CORRELATIONS:
    col = grid_mean[c]
    print(f"    correlation {c}:  {col.min():.1f}% to {col.max():.1f}%  (spread {col.max()-col.min():.1f} pts)")
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
data = grid_mean.to_numpy()
im = ax.imshow(data, cmap="Greys", aspect="auto", vmin=0, vmax=data.max())

ax.set_xticks(range(len(SWEEP_CORRELATIONS)))
ax.set_xticklabels(SWEEP_CORRELATIONS)
ax.set_yticks(range(len(SWEEP_CHANNEL_COUNTS)))
ax.set_yticklabels(SWEEP_CHANNEL_COUNTS)
ax.set_xlabel("Inter-channel correlation")
ax.set_ylabel("Number of channels in the plan")
ax.set_title("How much the average channel's estimate moves (CV %), no phasing")

threshold = data.max() * 0.55
for i in range(data.shape[0]):
    for j in range(data.shape[1]):
        ax.text(
            j, i, f"{data[i, j]:.0f}%",
            ha="center", va="center", fontweight="bold",
            color="white" if data[i, j] > threshold else "black",
        )

fig.colorbar(im, ax=ax, label="CV of recovered marginal return (%)")
fig.tight_layout()
plt.show()
```

**Finding.** Read the grid across and it climbs steeply; read it down and it
barely moves. At three channels, going from correlation 0.1 to 0.9 multiplies
the average channel's CV by roughly seven. Holding correlation at 0.7, going
from 3 channels to 20 moves it by about a tenth of that.

This reframes what Sections 1-3 measured. "Does it scale to 20 channels?" is
the wrong worry: what determines whether your model can tell channels apart is
how correlated they are, not how many there are.

The max-CV grid tells a different story on purpose, and it is worth
understanding why. There, channel count *does* appear to matter, roughly
doubling the worst channel's CV between K=3 and K=20 at high correlation. Most
of that is the max over K draws growing with K rather than any channel getting
genuinely worse, which is why the average is the honest summary for this
question. If your concern is specifically "my least reliable channel", the max
grid is the relevant one, and correlation still dominates it.


## Section 5 — Does the *fix* hold across the grid?

Section 4 swept correlation and channel count to size the problem, and found
correlation drives it while channel count barely matters. That says nothing
about whether the fix scales: it is perfectly possible for the damage to be
flat in channel count while phasing gets steadily worse at pulling it back,
since the on/off lever only frees up about one week a month per channel and
there is more to disentangle at 20 channels than at 3.

This section runs the same grid with the fix applied. For every cell: build a
208-week history and a 52-week plan at that correlation, measure the average
channel's CV on history + unphased plan, run `BudgetPhaser` with
`Blackout(max_dark_weeks_per_month=1)`, then re-measure on history + phased
plan. The cell value is the percentage reduction between the two.

Cheaper settings than Sections 2-3 (`grid_steps=12`, `n_phasing_seeds=6`,
3 draws) because each of the 25 cells runs its own grid search. The draw-to-draw
standard deviations printed below are small enough that this doesn't change any
conclusion, and they tighten as channel count rises.


```python
# ── Section 5: does phasing's benefit hold across the grid? ────────────────
SWEEP5_DRAWS = 3
SWEEP5_GRID_STEPS = 12
SWEEP5_PHASING_SEEDS = 6

improve_records = []
for n in SWEEP_CHANNEL_COUNTS:
    channels, marginal_returns = make_cycled_channels(n)
    for corr in SWEEP_CORRELATIONS:
        befores, afters, reductions = [], [], []
        for draw in range(SWEEP5_DRAWS):
            history = simulate_spend(
                n_obs=N_HISTORY, correlation=corr, channels=channels, seed=draw
            )
            history.index = pd.date_range("2019-01-07", periods=N_HISTORY, freq="W-MON")
            plan = simulate_spend(
                n_obs=PLAN_WEEKS, correlation=corr, channels=channels, seed=100 + draw
            )
            plan.index = pd.date_range(
                history.index[-1] + pd.Timedelta(weeks=1), periods=PLAN_WEEKS, freq="W-MON"
            )

            diag_before = CollinearityDiagnostic(
                spend_df=pd.concat([history, plan]), true_marginal_returns=marginal_returns
            ).fit(n_sims=N_SIMS)
            cv_before = diag_before.summary()["coef_of_variation"].mean()

            phaser = BudgetPhaser(
                history_df=history,
                plan_df=plan,
                true_marginal_returns=marginal_returns,
                max_weekly_deviation_pct=Blackout(max_dark_weeks_per_month=1),
            )
            phaser.fit(
                n_sims=N_SIMS,
                grid_steps=SWEEP5_GRID_STEPS,
                n_phasing_seeds=SWEEP5_PHASING_SEEDS,
            )

            diag_after = CollinearityDiagnostic(
                spend_df=pd.concat([history, phaser.recommended_schedule_]),
                true_marginal_returns=marginal_returns,
            ).fit(n_sims=N_SIMS)
            cv_after = diag_after.summary()["coef_of_variation"].mean()

            befores.append(cv_before)
            afters.append(cv_after)
            reductions.append(100 * (cv_before - cv_after) / cv_before)

        improve_records.append(
            {
                "n_channels": n,
                "correlation": corr,
                "cv_before_pct": 100 * float(np.mean(befores)),
                "cv_after_pct": 100 * float(np.mean(afters)),
                "reduction_pct": float(np.mean(reductions)),
                "reduction_std": float(np.std(reductions)),
            }
        )
    print(f"K={n:>2} done")

df_improve = pd.DataFrame(improve_records)
grid_reduction = df_improve.pivot(index="n_channels", columns="correlation", values="reduction_pct")
grid_before = df_improve.pivot(index="n_channels", columns="correlation", values="cv_before_pct")
grid_after = df_improve.pivot(index="n_channels", columns="correlation", values="cv_after_pct")

print("\n── Reliability gained %% (rows = channels, cols = correlation) ──")
print(grid_reduction.round(1))
print("\n── CV before phasing %% ──")
print(grid_before.round(1))
print("\n── CV after phasing %% ──")
print(grid_after.round(1))
```

```python
# ── What the fix grid says ─────────────────────────────────────────────────
print("── Key findings ──────────────────────────────────────────────")
col = grid_reduction[0.7]
print(f"  At correlation 0.7, reliability gained by channel count:")
for n in SWEEP_CHANNEL_COUNTS:
    sd = df_improve[(df_improve.n_channels == n) & (df_improve.correlation == 0.7)]["reduction_std"].iloc[0]
    print(f"    K={n:>2}: {col[n]:.1f}%  (+/-{sd:.1f} across draws)")
print(f"  → range across channel counts: {col.max()-col.min():.1f} points. The fix does not decay with channel count.")
print()
print("  Reliability gained rises with correlation (K=3 row):")
print("    " + "  ".join(f"r={c}: {grid_reduction.loc[3, c]:.1f}%" for c in SWEEP_CORRELATIONS))
print("  → phasing does most where there is most to fix.")
print()
print("  And the low numbers on the left are not a failure. CV before phasing there:")
print("    " + "  ".join(f"r={c}: {grid_before.loc[3, c]:.1f}%" for c in SWEEP_CORRELATIONS))
print(f"  → after phasing, every cell in the grid lands between "
      f"{grid_after.to_numpy().min():.1f}% and {grid_after.to_numpy().max():.1f}%, wherever it started.")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, data, title, cmap in (
    (axes[0], grid_reduction.to_numpy(), "Reliability gained from phasing (%)", "Greens"),
    (axes[1], grid_after.to_numpy(), "CV after phasing (%)", "Greys"),
):
    im = ax.imshow(data, cmap=cmap, aspect="auto", vmin=0, vmax=data.max())
    ax.set_xticks(range(len(SWEEP_CORRELATIONS)))
    ax.set_xticklabels(SWEEP_CORRELATIONS)
    ax.set_yticks(range(len(SWEEP_CHANNEL_COUNTS)))
    ax.set_yticklabels(SWEEP_CHANNEL_COUNTS)
    ax.set_xlabel("Inter-channel correlation")
    ax.set_ylabel("Number of channels in the plan")
    ax.set_title(title)
    thr = data.max() * 0.55
    for i in range(data.shape[0]):
        for j in range(data.shape[1]):
            ax.text(j, i, f"{data[i, j]:.0f}%", ha="center", va="center",
                    fontweight="bold", color="white" if data[i, j] > thr else "black")
    fig.colorbar(im, ax=ax)

fig.tight_layout()
plt.show()
```

**Finding.** The fix does not decay with channel count. At a realistic 0.7
correlation the on/off lever buys about 50-53% whether the plan carries 3
channels or 20, and the spread across channel counts is a couple of points,
inside the draw-to-draw noise. The concern that 20 channels is simply too many
to disentangle with one dark week a month does not show up in the numbers.

Two things worth reading off this grid beyond the headline:

1. **Reliability gained rises with correlation**, from around 12-19% at 0.1 to
   around 70-74% at 0.9. Phasing does most where there is most to fix, which is
   the right shape for a diagnostic-led tool.
2. **The modest gains on the left are not a weakness.** At correlation 0.1 the
   average channel's estimate only moves 4.6% before any phasing, so there is
   very little left to win. The `CV after phasing` grid makes this concrete:
   every cell lands between roughly 4% and 12% regardless of where it started.
   Phasing pulls the whole grid down to a similar floor.
